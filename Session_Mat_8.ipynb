```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

users = [1200, 1500, 1400, 1800, 2000, 2500, 2200]
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [20, 25, 30, 28, 35, 40, 45]

swiggy = [18, 22, 27, 32, 30, 38, 42]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(days, zomato, linestyle="--", marker="o", label="Zomato") # linestyle="--"  =>  Dashed line
plt.plot(days, swiggy, linestyle=":", marker="s", label="Swiggy")# linestyle=":" => Dotted line
plt.legend() #Dono lines ke naam show karta hai
```

```python
days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
        "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]

sales = [12000, 15000, 13000, 18000, 20000,
         17000, 22000, 25000, 23000, 28000]
```

```python
plt.style.use("ggplot") #ggplot Matplotlib ki ek built-in style hai jo graph ka overall look change karti hai.
plt.figure(figsize=(10, 5))
plt.plot(days, sales, color="orange", marker="o")
plt.title("Flipkart Daily Sales")
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

posts = [10, 15, 12, 20, 18, 25, 22]
colors = [
    "#833AB4",
    "#C13584",
    "#E1306C",
    "#FD1D1D",
    "#F77737",
    "#FCAF45",
    "#833AB4"]
#833AB4 => Purple
#C13584 => Pink-purple
#E1306C => Pink
#FD1D1D => Red
#F77737 => Orange
#FCAF45 => Yellow-orange
```

```python
plt.bar(days, posts, color=colors)
plt.title("Instagram Daily Posts")  #  title() => Graph ka title.
```
