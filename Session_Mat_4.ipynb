```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#                                                                -->   Task-1     <--
```

```python
apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
hours = [3, 2, 1, 1, 1]
```

```python
plt.pie(hours,labels=apps,autopct="%1.1f%%")
#labels=apps =>Chart par app ke naam dikhata hai
#autopct="%1.1f%%" =>Har slice par percentage show karta hai
```

```python
#                                                                -->   Task-2     <--
```

```python
categories = ["Food Delivery","Shopping","Entertainment","UPI Payments"]
spending = [4000, 6000, 2000, 3000]
explode = [0, 0.2, 0, 0] #explode = [0, 0.2, 0, 0] =>Is list mein har number ek category ke slice ko control karta hai.
```

```python
plt.pie(spending,labels=categories,autopct="%1.1f%%",explode=explode)
#autopct =>  Percentage show karta hai.
```

```python
categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]
spending = [4000, 6000, 2000, 3000]
```

```python
fig, ax = plt.subplots()
ax.pie(spending,labels=categories,autopct="%1.1f%%")
circle = plt.Circle((0, 0),0.5,color="white")
ax.add_artist(circle)
```
