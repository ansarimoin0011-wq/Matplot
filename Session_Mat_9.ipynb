```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import matplotlib.dates as mdates
```

```python
#                                                                -->   Task-1     <--
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [20, 25, 30, 28, 35, 40, 45]

swiggy = [15, 20, 25, 22, 30, 35, 40]
```

```python
plt.bar(days, zomato, label="Zomato", color="orange")
plt.bar(days, swiggy, bottom=zomato, label="Swiggy", color="green")
bottom=zomato  #  Swiggy ka bar Zomato ke bar ke upar se start hoga.
plt.legend() 
plt.title("Daily Food Orders - Zomato vs Swiggy")
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [1200, 1400, 1300, 1600, 1800, 2100, 2000]

youtube = [1800, 2000, 1900, 2200, 2400, 2700, 2500]
```

```python
plt.fill_between(days, spotify, color="green", alpha=0.5, label="Spotify") 
plt.fill_between(days, youtube, color="red", alpha=0.5, label="YouTube")
plt.plot(days, spotify, color="green")
plt.plot(days, youtube, color="red")
plt.title("Daily Active Users - Spotify vs YouTube")
#  fill_between() => line ke neeche area fill karta hai.
#  alpha=0.5 => color ko thoda transparent banata hai.
#  label="Spotify" => Legend ke liye naam.
```

```python
days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
        "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]
delivery_time = [30, 32, 28, 35, 31, 29, 34, 36, 33, 30]
std = [3, 4, 2, 5, 3, 2, 4, 3, 4, 2]
```

```python
plt.errorbar(days,delivery_time,yerr=std,fmt="o-",capsize=5) #yerr => Error/standard deviation
plt.title("Flipkart Average Delivery Time with Standard Deviation")
```

```python
start_date = datetime(2026, 9, 1)
dates = [start_date + timedelta(days=i) for i in range(14)]
balance = [2500, 2700, 2400, 3000, 2800, 3200, 3100,
           3500, 3300, 3600, 3400, 3800, 4000, 3900]
```

```python
plt.figure(figsize=(12, 5))
plt.plot(dates, balance, marker="o")
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
plt.xticks(rotation=45)  #Isse dates thodi tilted ho jayengi aur easily read hongi.
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [1200, 1400, 1300, 1600, 1800, 2100, 2000]

youtube = [1800, 2000, 1900, 2200, 2400, 2700, 2500]
```
