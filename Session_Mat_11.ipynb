```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#                                                                -->   Task-1     <--
```

```python
restaurants = ["Dominos","La Pino'z","The Pizza","Burger King","Subway"]
orders = [120, 95, 150, 110, 80]
```

```python
plt.figure(figsize=(10, 5))
plt.bar(restaurants, orders)
plt.savefig("zomato_orders.png", dpi=150)
plt.show()
```

```python
# Create data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6200, 5800, 7000, 8200, 9500, 7800]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(days, steps, marker="o")
plt.title("Daily Step Count")
plt.savefig("daily_steps.svg")
plt.show()
#Main learning: plt.savefig() ke filename ki extension change karke chart ko different formats mein save kar sakte ho.
```

```python
apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_spent = [30, 35, 20, 15]
```

```python
plt.figure(figsize=(7, 7))
plt.pie(time_spent,labels=apps,autopct="%1.1f%%")
plt.title("Daily App Time Spending")
plt.savefig("app_time_spending.pdf")# file ko Pdf me save krna
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 180, 150, 220, 200]
```

```python
plt.figure(figsize=(10, 5))

plt.bar(months, sales)
plt.title("Monthly Sales")
plt.savefig("monthly_sales_300dpi.pdf", dpi=300)
```
