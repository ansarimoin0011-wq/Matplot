```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_hours = [1, 2, 1.5, 2.5, 3, 4, 2]
youtube_hours = [2, 1.5, 3, 2, 4, 5, 3.5]
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5)) 
ax[0].plot(days, spotify_hours, marker="o")

ax[0].set_title("Daily Spotify Listening Hours")
ax[0].set_xlabel("Day")
ax[0].set_ylabel("Hours")
ax[1].plot(days, youtube_hours, marker="o", color="orange")

ax[1].set_title("Daily YouTube Viewing Hours")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Hours")
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
spotify_hours = [1, 2, 1.5, 2.5, 3, 4, 2]
youtube_hours = [2, 1.5, 3, 2, 4, 5, 3.5]

fig, ax = plt.subplots(1, 2, figsize=(12, 5)) 
ax[0].plot(days, spotify_hours, marker="o")

ax[0].set_title("Daily Spotify Listening Hours")
ax[0].set_xlabel("Day")
ax[0].set_ylabel("Hours")
ax[1].plot(days, youtube_hours, marker="o", color="orange")

ax[1].set_title("Daily YouTube Viewing Hours")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Hours")


ax[0].set_title("Daily Spotify Listening Hours") 
ax[1].set_title("Daily YouTube Viewing Hours")
#Meaning: Dono graphs ko alag-alag title dena.

ax[0].set_xlabel("Day")
ax[0].set_ylabel("Listening Hours")

ax[1].set_xlabel("Day")
ax[1].set_ylabel("Viewing Hours")
ax[0].grid(True) #grid(True) graph ke andar grid lines show karta hai, jisse values ko read karna easy hota hai.
ax[1].grid(True)
```
