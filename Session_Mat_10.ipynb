```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

users = [
    1200, 1400, 1350, 1600, 1800, 1750,
    2000, 2200, 2100, 2300, 2500, 2400
]
```

```python
plt.figure(figsize=(12, 5))
plt.plot(months, users, marker="o")
max_users = max(users)
max_index = users.index(max_users)
plt.annotate("Highest Users",xy=(months[max_index], users[max_index]),xytext=("Sep", 2700),arrowprops=dict(arrowstyle="->"))
plt.title("Monthly Active Users - Music Streaming App")
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"]

orders = [450, 520, 380, 600, 550, 480, 650]
```

```python
plt.figure(figsize=(10, 5))
plt.bar(months, orders)
min_orders = min(orders)
min_index = orders.index(min_orders)  # Step 4: Find position of lowest value
plt.annotate("Lowest Orders",xy=(months[min_index], orders[min_index]),xytext=("Jan", 700),arrowprops=dict(arrowstyle="->")) 
#Ye batata hai ki "Lowest Orders" text kahan dikhana hai.
#Ye arrow ka target point hai.
plt.title("Monthly Food Orders - Zomato")
```

```python
payment_methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [45, 30, 15, 10]
```

```python
plt.figure(figsize=(7, 7))
plt.pie(payments,labels=payment_methods,autopct="%1.1f%%")
plt.title("Flipkart Payment Methods")
plt.legend(loc="lower right")
```
