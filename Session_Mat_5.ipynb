```python
np.random.seed(42)
scores = np.random.randint(120, 221, 50)
plt.figure(figsize=(15, 4))
plt.subplot(1, 3, 1)
plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("IPL Scores - 5 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
```

```python
plt.subplot(1, 3, 2)
plt.hist(scores, bins=10, color="orange", edgecolor="black")
plt.title("IPL Scores - 10 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
```

```python
plt.subplot(1, 3, 3)
plt.hist(scores, bins=15, color="lightgreen", edgecolor="black")
plt.title("IPL Scores - 15 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
```

```python
steps = [
    4500, 6000, 7500, 8000, 5500, 9000, 7000, 6500, 10000, 8500,
    5000, 7500, 11000, 9500, 6000, 8000, 7000, 12000, 9000, 5500,
    8500, 10000, 6500, 7500, 9500, 11000, 8000, 6000, 9000, 7000
]
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
                                                                # First histogram: Frequency
plt.hist(steps, bins=8, color="skyblue", edgecolor="black")
plt.title("Daily Steps - Frequency")
plt.xlabel("Number of Steps")
plt.ylabel("Frequency (Number of Days)")
plt.subplot(1, 2, 2)
                                                                   # Second histogram: Density
plt.hist(steps, bins=8, density=True, color="orange", edgecolor="black")
plt.title("Daily Steps - Density")
plt.xlabel("Number of Steps")
plt.ylabel("Density")
plt.tight_layout()  # dono chart ko sath me show karn ke liye 
```

```python
# Y-axis = Number of Days
# Har bar batayega ki us steps range mein kitne din aaye.
# Example: Agar kisi group mein 6 din hain, toh bar ki height 6 hogi.

# Y-axis = Density
# Bar ki height normalized density dikhati hai. Ismein bar ki height ka matlab days ki direct count nahi hai.
The frequency histogram shows the number of days in each step-count group. 
The density histogram shows the normalized distribution of the data. 
The total area under the density histogram is 1.
Therefore, the Y-axis values are different in both histograms.
```
