```python
influencers = [
    "Influencer A",
    "Influencer B",
    "Influencer C",
    "Influencer D",
    "Influencer E",
    "Influencer F"
]
followers = [12, 8, 15, 10, 20, 6]
```

```python
plt.barh(influencers, followers, color="skyblue")
# plt.barh( )=>Horizontal bar chart banata hai
#color="skyblue" =>Bars ka color sky blue karta hai
plt.title("Instagram Influencers Followers")
plt.xlabel("Followers (in millions)")
```

```python
apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 100, 80, 60, 90]
```

```python
bars = plt.bar(apps,orders,color=["red", "orange", "blue", "green", "purple"])
for bar in bars:
    height = bar.get_height()
plt.text(bar.get_x() + bar.get_width() / 2,height,str(int(height)),ha="center",va="bottom")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")
```
