```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#                                                                -->   Task-1     <--
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [4500, 6200, 5800, 7500, 6800, 9000, 7200]

water = [2.0, 2.5, 2.2, 3.0, 2.8, 3.2, 2.7]
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 8))
#                                                                  │   Graph 1: Steps        │ 
ax[0].plot(days, steps, marker="o")
ax[0].set_title("Daily Step Count") #Ye graph ko labels deta hai.
#                                                                  │   Graph 2: Water Intake │
ax[1].plot(days, water, marker="o")
ax[1].set_title("Daily Water Intake") # #Ye graph ko labels deta hai.
plt.tight_layout() #Isse dono graphs ke beech proper spacing aa jaati hai.
```

```python
quarters = ["Q1", "Q2", "Q3", "Q4"]

posts = [30, 45, 60, 50]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].set_title("Instagram Posts - Line Chart") #ax[0, 0] =>Top-left graph
ax[0, 0].plot(quarters, posts, marker="o") #

ax[0, 1].set_title("Instagram Posts - Bar Chart")#  ax[0, 1]=> Top-right graph
ax[0, 1].bar(quarters, posts, color="orange")

ax[1, 0].set_title("Instagram Posts - Scatter Plot")#   ax[1, 0]=> Bottom-left graph
ax[1, 0].scatter(quarters, posts, color="green")

ax[1, 1].set_title("Instagram Posts - Pie Chart")#  ax[1, 1]=>Bottom-right graph
ax[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
zomato = [12, 15, 18, 14, 20, 22]
swiggy = [10, 13, 16, 19, 17, 21]
dominos = [8, 11, 14, 12, 16, 18]
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 5), sharey=True) #sharey=True ka matlab =>Teeno graphs ka Y-axis same scale use karega.

ax[0].set_title("Zomato Orders")#ax[0] => Pehla graph.
ax[0].plot(months, zomato, marker="o", color="red")

ax[1].plot(months, swiggy, marker="o", color="orange")#ax[1] => Doosra graph.
ax[1].set_title("Swiggy Orders")

ax[2].plot(months, dominos, marker="o", color="blue")#ax[2] => Teesra graph.
ax[2].set_title("Domino's Orders")
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
youtube = [5, 6, 8, 7, 9, 10]
instagram = [3, 4, 5, 6, 5, 7]
spotify = [1, 2, 2, 3, 3, 4]
whatsapp = [1, 1.5, 2, 2, 2.5, 3]
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8), sharex=True)

ax[0, 0].plot(months, youtube, marker="o", color="red")
ax[0, 0].set_title("YouTube Data Usage")

ax[0, 1].plot(months, instagram, marker="o", color="blue")
ax[0, 1].set_title("Instagram Data Usage")

ax[1, 0].plot(months, spotify, marker="o", color="green")
ax[1, 0].set_title("Spotify Data Usage")

ax[1, 1].plot(months, whatsapp, marker="o", color="orange")
ax[1, 1].set_title("WhatsApp Data Usage")
```
