```python
#                                                                -->   Task-2     <--
```

```python
sales_df = pd.read_csv("sales.csv")
category_sales = sales_df.groupby("category")["sales_amount"].sum()
plt.figure(figsize=(8, 5))
plt.bar(category_sales.index,category_sales.values)
plt.title("Total Sales by Product Category")
```

```python
#                                                                -->   Task-3     <--
```

```python
sales_df = pd.read_csv("sales.csv") # Load sales data
sales_df["date"] = pd.to_datetime(sales_df["date"])# Convert date column to datetime
sales_df["month"] = sales_df["date"].dt.to_period("M")  # Create month column
monthly_sales = sales_df.groupby("month")["sales_amount"].sum()# Calculate monthly total sales
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index.astype(str),monthly_sales.values,marker="o")
plt.title("Monthly Sales Trend")
plt.grid(True)
plt.tight_layout()
```

```python
sales_df = pd.read_csv("sales.csv")
plt.figure(figsize=(10, 5))
plt.hist(sales_df["sales_amount"],bins=10)
plt.title("Distribution of Sales Amounts")
plt.grid(axis="y", alpha=0.3)  # Add grid
```

```python
sales_df = pd.read_csv("sales.csv") # Load sales data
category_sales = sales_df.groupby("category")["sales_amount"].sum()# Calculate total sales by category
plt.figure(figsize=(10, 5))
plt.bar(category_sales.index,category_sales.values)
for i, value in enumerate(category_sales.values):# Add exact values above each bar
    plt.text(i,value,str(value),ha="center",va="bottom")
plt.title("Category-wise Total Sales")
plt.savefig("category_sales_kpi.png", dpi=150)
```
